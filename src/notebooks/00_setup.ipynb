# 00 · Setup do Unity Catalog

Cria o catálogo, os schemas de cada camada e o Volume da landing zone.

**Rode este notebook uma única vez, antes de qualquer outro.**

Se a sua conta não tiver permissão para `CREATE CATALOG`, troque a constante
`CATALOG` em `src/config.py` para um catálogo já existente (no Free Edition,
`workspace` funciona) e rode apenas os comandos de schema e volume.

```python
# Coloca a raiz do repositório no sys.path.
# Sobe diretórios até achar a pasta `src`, de modo que o notebook funcione
# tanto em Git Folders quanto em Workspace Files, sem caminho hardcoded.
import os
import sys

_root = os.getcwd()
while _root != "/" and not os.path.isdir(os.path.join(_root, "src")):
    _root = os.path.dirname(_root)
if _root not in sys.path:
    sys.path.insert(0, _root)
print("Repo root:", _root)
```

```python
from src import config

print('Catálogo :', config.CATALOG)
print('Landing  :', config.LANDING_ROOT)
```

```python
%sql
CREATE CATALOG IF NOT EXISTS ifood_case
COMMENT 'Case técnico Data Architect - corridas de táxi de NY';
```

```python
%sql
CREATE SCHEMA IF NOT EXISTS ifood_case.landing COMMENT 'Arquivos originais da TLC, sem transformação';
CREATE SCHEMA IF NOT EXISTS ifood_case.raw     COMMENT 'Delta espelhando a origem, tudo em string';
CREATE SCHEMA IF NOT EXISTS ifood_case.trusted COMMENT 'Camada de consumo: colunas do case, já tipadas';
CREATE SCHEMA IF NOT EXISTS ifood_case.refined COMMENT 'Fato unificado da frota e agregados analíticos';
CREATE SCHEMA IF NOT EXISTS ifood_case.quality COMMENT 'Resultado das regras de qualidade de dados';
```

```python
%sql
CREATE VOLUME IF NOT EXISTS ifood_case.landing.files
COMMENT 'Landing zone: parquets originais publicados pela NYC TLC';
```
